```python
columns_to_keep = ['Age', 'Gender', 'Marital Status', 'Occupation', 'Monthly Income',
       'Educational Qualifications', 'Family size','Pin code','Output']
columns_to_drop = ['latitude', 'longitude','Feedback', 'Unnamed: 12']
```

```python
modify_data = data[columns_to_keep]
```

```python
data.drop(columns=columns_to_drop, inplace=True)
```

```python
data.head(1)
```

```python
data.rename(columns = {"Monthly Income" : "Monthly Salary"},inplace=True)
```

```python
data.head(1)
```

```python
data.duplicated().sum()
```

```python
data.isna().sum()
```

```python
data.dropna(inplace =True)
```

```python
data.isna().sum()
```

```python
data["Age (Month)"] = data["Age"]
```

```python
data.head(1)
```

```python
data["Gender"] = data["Gender"].apply(lambda x : "F" if x == "Female" else "M")
```

```python
data["Family size"] = data["Family size"].apply(lambda x:"Large" if x>3 else "Small" )
```

```python
data
```

```python
data["Output"] = data["Output"].apply(lambda x:1 if x == "Yes" else 0)
```

```python
data
```

```python
data["Output"] = data["Output"].astype(float)
```
